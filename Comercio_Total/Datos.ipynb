```python
import comtradeapicall
import pandas as pd

pares = [
    #CA_USA AND WORLD
    {'reporter': '124', 'partner': '842', 'flow': 'X', 'label': 'CA_USA'},
    {'reporter': '124' , 'partner': '842', 'flow': 'M', 'label': 'USA_CA'},
    {'reporter': '124', 'partner': '0', 'flow': 'X', 'label': 'CA_W'},
    {'reporter': '124' , 'partner': '0', 'flow': 'M', 'label': 'W_CA'},
    #CA_MX 
    {'reporter': '124', 'partner': '484', 'flow': 'X', 'label': 'CA_MX'},
    {'reporter': '124' , 'partner': '484', 'flow': 'M', 'label': 'MX_CA'},
    #MX_USA AND WORLD
    {'reporter': '484', 'partner': '842', 'flow': 'X', 'label': 'MX_USA'},
    {'reporter': '484' , 'partner': '842', 'flow': 'M', 'label': 'USA_MX'},

    {'reporter': '484', 'partner': '0', 'flow': 'X', 'label': 'MX_W'},
    {'reporter': '484' , 'partner': '0', 'flow': 'M', 'label': 'W_MX'},

    #USA_WORLD
    {'reporter': '842', 'partner': '0', 'flow': 'X', 'label': 'USA_W'},
    {'reporter': '842' , 'partner': '0', 'flow': 'M', 'label': 'W_USA'},

]

frames = []

for par in pares:
    for start in range(1994, 2026, 12):
        end = min(start + 11, 2025)
        years = ','.join(str(y) for y in range(start, end + 1))
        
        df = comtradeapicall.previewFinalData(
            typeCode='C', freqCode='A', clCode='HS',
            period=years,
            reporterCode=par['reporter'],
            cmdCode='TOTAL',
            flowCode=par['flow'],
            partnerCode=par['partner'],
            partner2Code=None, customsCode=None, motCode=None,
            maxRecords=500, format_output='JSON',
            aggregateBy=None, breakdownMode='classic',
            countOnly=None, includeDesc=True
        )
        
        if df is not None and not df.empty:
            df['par'] = par['label']
            frames.append(df)

final = pd.concat(frames, ignore_index=True)
final['netWgt_ton'] = final['netWgt'] / 1000
final.rename(columns={'period':'Año', 'primaryValue': 'Valor Total (USD)', 'netWgt_ton':'Cantidad Total (toneladas)'}, inplace=True)
cols = ['Año', 'par', 'reporterDesc', 'partnerDesc', 'cmdCode', 'cmdDesc', 'Valor Total (USD)']
# Guardar un archivo por cada flow
for label, grupo in final.groupby('par'):
    nombre = f'TOTAL_{label}.csv'
    grupo[cols].to_csv(nombre, index=False)
    print(f'Guardado: {nombre} — {len(grupo)} filas')
```
