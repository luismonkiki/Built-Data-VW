# Intensidad comercial

```python
import pandas as pd

# Exportaciones bilaterales TMEC
Xij_CA = TCM.set_index('Año')['Valor Total (USD)']  + TCU.set_index('Año')['Valor Total (USD)']
Xij_MX = TMC.set_index('Año')['Valor Total (USD)'] + TMU.set_index('Año')['Valor Total (USD)']  # MX → CA+USA  
Xij_USA = TUC.set_index('Año')['Valor Total (USD)'] + TUM.set_index('Año')['Valor Total (USD)']  # USA → CA+MX

# Exportaciones totales (al mundo)
Xtot_CA  = TCW.set_index('Año')['Valor Total (USD)']
Xtot_MX  = TMW.set_index('Año')['Valor Total (USD)']
Xtot_USA = TUW.set_index('Año')['Valor Total (USD)']

# Importaciones bilaterales TMEC
Mij_CA  = TMC.set_index('Año')['Valor Total (USD)'] + TUC.set_index('Año')['Valor Total (USD)']
Mij_MX  = TCM.set_index('Año')['Valor Total (USD)'] + TUM.set_index('Año')['Valor Total (USD)']
Mij_USA = TCU.set_index('Año')['Valor Total (USD)'] + TMU.set_index('Año')['Valor Total (USD)']

# Importaciones totales (del mundo)
Mtot_CA  = TWC.set_index('Año')['Valor Total (USD)']
Mtot_MX  = TWM.set_index('Año')['Valor Total (USD)']
Mtot_USA = TWU.set_index('Año')['Valor Total (USD)']

# Alinear años comunes
años = Xij_CA.index.intersection(Xij_MX.index).intersection(Xij_USA.index)\
              .intersection(Xtot_CA.index).intersection(Xtot_MX.index).intersection(Xtot_USA.index)\
              .intersection(Mtot_CA.index).intersection(Mtot_MX.index).intersection(Mtot_USA.index)

# Sumas TMEC
sum_Xij   = Xij_CA[años]  + Xij_MX[años]  + Xij_USA[años]
sum_Xtot  = Xtot_CA[años] + Xtot_MX[años] + Xtot_USA[años]
sum_Mij   = Mij_CA[años]  + Mij_MX[años]  + Mij_USA[años]
sum_Mtot  = Mtot_CA[años] + Mtot_MX[años] + Mtot_USA[años]

# Mundiales (exportaciones e importaciones globales)
XMundiales = sum_Xtot  # proxy: suma exportaciones totales TMEC al mundo
MMundiales = sum_Mtot

# Fórmula 21 - Intensidad exportaciones
int_exp = (sum_Xij / sum_Xtot) / (sum_Xtot / XMundiales)

# Fórmula 22 - Intensidad importaciones
int_imp = (sum_Mij / sum_Mtot) / (sum_Mtot / MMundiales)

# Fórmula 23 - Intensidad comercial
int_com = ((sum_Xij + sum_Mij) / (sum_Xtot + sum_Mtot)) / \
          ((sum_Xtot + sum_Mtot) / (XMundiales + MMundiales))

# Resultado
df_intensidad = pd.DataFrame({
    'Año': años,
    'Intensidad Exportaciones': int_exp.values,
    'Intensidad Importaciones': int_imp.values,
    'Intensidad Comercial': int_com.values
})

print(df_intensidad)
    
```
